```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Create widgets for the chat UI
text_input = widgets.Text(
    value='',
    placeholder='Ask a question about RAG or Ollama...',
    description='Question:',
    layout=widgets.Layout(width='70%')
)

send_button = widgets.Button(
    description='Send',
    button_style='success',
    tooltip='Click to ask the RAG chatbot',
    layout=widgets.Layout(width='15%')
)

clear_button = widgets.Button(
    description='Clear Chat',
    button_style='warning',
    tooltip='Clear the conversation history',
    layout=widgets.Layout(width='15%')
)

output_area = widgets.Output(
    layout=widgets.Layout(border='1px solid #ccc', padding='10px', height='300px', overflow='y-scroll')
)

def on_send_clicked(b):
    question = text_input.value.strip()
    if not question:
        return

    # Disable input while processing
    text_input.disabled = True
    send_button.disabled = True

    with output_area:
        print(f"\n🧑 User: {question}")
        print("🤖 AI is thinking...")

        try:
            # Call the rag_chat function previously defined
            answer = rag_chat(question)
            # Clear the 'thinking' line and print the actual response
            clear_output(wait=True)
            print(f"\n🧑 User: {question}")
            print(f"🤖 AI: {answer}")
        except Exception as e:
            clear_output(wait=True)
            print(f"\n🧑 User: {question}")
            print(f"❌ Error: Could not reach Ollama. Make sure Ollama is running. ({str(e)})")

    # Reset input field and enable controls
    text_input.value = ''
    text_input.disabled = False
    send_button.disabled = False

def on_clear_clicked(b):
    with output_area:
        clear_output()
    print("Chat cleared!")

# Bind actions to buttons and text submit event
send_button.on_click(on_send_clicked)
text_input.on_submit(on_send_clicked)
clear_button.on_click(on_clear_clicked)

# Lay out the interface
input_box = widgets.HBox([text_input, send_button, clear_button])
chat_interface = widgets.VBox([output_area, input_box])

# Display the interactive chatbox
print("=== RAG local Chatbot Interface ===")
display(chat_interface)

```

```python
from sentence_transformers import SentenceTransformer

# Load the embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# Convert document chunks into embeddings
embeddings = embedding_model.encode(chunks)

# Display results
print("Embeddings generated successfully!")
print("Embedding shape:", embeddings.shape)

print("\nFirst chunk embedding:")
print(embeddings[0])
```

```python
import numpy as np

def retrieve(question, top_k=2):

    # Convert question into an embedding
    question_embedding = embedding_model.encode([question])

    # Calculate cosine similarity
    similarities = np.dot(
        embeddings,
        question_embedding[0]
    ) / (
        np.linalg.norm(embeddings, axis=1) *
        np.linalg.norm(question_embedding[0])
    )

    # Find the most relevant chunks
    best_indices = np.argsort(similarities)[::-1][:top_k]

    return [chunks[i] for i in best_indices]


# Test retrieval
question = "What is the purpose of embeddings?"

results = retrieve(question)

for i, result in enumerate(results):
    print(f"\nRetrieved Chunk {i+1}:")
    print(result)
```

```python
import requests

def rag_chat(question):

    # Step 1: Retrieve relevant document chunks
    results = retrieve(question)

    # Step 2: Combine retrieved chunks
    context = "\n".join(results)

    # Step 3: Prepare the prompt
    prompt = f"""
You are a helpful AI assistant.

Answer the user's question using ONLY the information
provided in the context.

If the answer is not available in the context,
say "I don't have enough information in the document."

CONTEXT:
{context}

QUESTION:
{question}

ANSWER:
"""

    # Step 4: Send prompt to Ollama
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": "llama3.2:1b",
            "prompt": prompt,
            "stream": False
        },
        timeout=180
    )

    response.raise_for_status()

    # Step 5: Extract AI response
    answer = response.json()["response"]

    return answer


# Test our RAG chatbot
question = "What is the purpose of embeddings?"

answer = rag_chat(question)

print("USER QUESTION:", question)
print("\nAI ANSWER:", answer)
```
